**What is OCR (Optical Character Recognition)?**
Optical Character Recognition (OCR) is a technology that scans printed or handwritten text from images, PDFs, or documents and converts it into editable and searchable digital text. Instead of manually typing the content, OCR automatically recognizes the characters and extracts the text. OCR is widely used for digitizing books, forms, invoices, historical documents, and other paper-based records.

**Why is Urdu OCR harder than English OCR?**
Urdu OCR is more challenging than English OCR because Urdu is written in a cursive script where most letters are connected to each other. Many Urdu letters have very similar shapes and differ only by the number or position of dots, making them difficult to recognize accurately. In addition, different fonts, handwriting styles, image quality, and complex document layouts make Urdu OCR more difficult than English OCR. OCR systems usually perform preprocessing, segmentation, and character recognition to improve the accuracy of Urdu text extraction.

**What are 2 real-world situations where Urdu OCR would be useful?**
Urdu OCR is very useful for visually impaired people because it can extract text from books or documents and convert it into speech using text-to-speech technology. It is also useful for government offices, libraries, and legal organizations where thousands of Urdu documents, gazettes, and official records need to be digitized and stored in searchable electronic form. This saves time, reduces manual work, and makes documents easier to search and manage.

```python
import csv
import os

labels = []


selected = df.head(50)

for _, row in selected.iterrows():

    filename = os.path.basename(row["file_name"])

    labels.append({
        "image": f"data/raw/dataset/{filename}",
        "text": row["text"]
    })



news_folder = "/content/data/raw/newspaper/screenshots"

for img in sorted(os.listdir(news_folder))[:30]:

    labels.append({
        "image": f"data/raw/newspaper/{img}",
        "text": "WRITE_THE_URDU_TEXT_HERE"
    })




synthetic_folder = "/content/data/raw/synthetic/synthetic"

for img in sorted(os.listdir(synthetic_folder)):

    labels.append({
        "image": f"data/raw/synthetic/{img}",
        "text": "WRITE_THE_URDU_TEXT_HERE"
    })

with open("/content/data/labels.csv","w",newline="",encoding="utf-8") as f:

    writer = csv.DictWriter(f,fieldnames=["image","text"])

    writer.writeheader()

    writer.writerows(labels)

print(len(labels))
print("labels.csv created successfully!")
```
